```python
import re
import nltk
from nltk.corpus import stopwords

# Download stopwords
nltk.download('stopwords')
```

```python
documents = {
    'doc1': "The quick brown fox jumps over the lazy dog. The dog is lazy.",
    'doc2': "A brown cat sat on the mat. The cat is quick.",
    'doc3': "Foxes are clever. Dogs are loyal. Cats are independent.",
    'doc4': "Quick dogs and lazy cats are good pets."
}

print("Sample Documents:\n")

for doc_id, content in documents.items():
    print(doc_id, ":", content)
```

```python
stop_words = set(stopwords.words('english'))

def preprocess_text(text):

    # Convert text to lowercase and extract words
    words = re.findall(r'\b\w+\b', text.lower())

    # Remove stop words
    filtered_words = []

    for word in words:
        if word not in stop_words:
            filtered_words.append(word)

    return filtered_words


# Preprocess all documents
processed_documents = {}

for doc_id, content in documents.items():
    processed_documents[doc_id] = preprocess_text(content)


print("Processed Documents:\n")

for doc_id, words in processed_documents.items():
    print(doc_id, ":", words)
```

```python
inverted_index = {}

for doc_id, words in processed_documents.items():

    for word in words:

        if word not in inverted_index:
            inverted_index[word] = []

        if doc_id not in inverted_index[word]:
            inverted_index[word].append(doc_id)


print("Inverted Index:\n")

for word, doc_ids in inverted_index.items():
    print(word, ":", doc_ids)
```

```python
def search_documents(query):

    # Preprocess the search query
    query_words = preprocess_text(query)

    print("Processed Query:", query_words)

    # If query is empty
    if not query_words:
        print("No valid search terms found.")
        return

    # Find documents containing the first word
    result = set(inverted_index.get(query_words[0], []))

    # Find common documents for remaining words
    for word in query_words[1:]:
        result = result.intersection(inverted_index.get(word, []))

    # Display retrieved documents
    if result:
        print("\nRetrieved Documents:\n")

        for doc_id in sorted(result):
            print(doc_id, ":", documents[doc_id])

    else:
        print("\nNo documents found.")
```

```python
# Test Case 1
print("Search Query: quick brown fox")
search_documents("quick brown fox")

print("\n-----------------------------\n")

# Test Case 2
print("Search Query: lazy cats")
search_documents("lazy cats")

print("\n-----------------------------\n")

# Test Case 3
print("Search Query: elephant")
search_documents("elephant")
```
