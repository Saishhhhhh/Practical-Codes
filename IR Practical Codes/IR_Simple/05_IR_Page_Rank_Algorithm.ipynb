```python
# IMPLEMENTATION OF PAGERANK ALGORITHM

import numpy as np

# 1. CREATE A WEB GRAPH

graph = {
    "A": ["B", "C"],
    "B": ["C", "D"],
    "C": ["A"],
    "D": ["C", "E"],
    "E": ["A", "D"]
}

print("Web Graph:")

for page in graph:
    print(page, "->", graph[page])
```

```python
# 2. INITIALIZE PAGE RANK

pages = list(graph.keys())

n = len(pages)

# Give equal rank to all pages
rank = np.ones(n) / n

# Damping factor
d = 0.85

# Number of iterations
iterations = 100
```

```python
# 3. IMPLEMENT PAGERANK ALGORITHM

for i in range(iterations):

    new_rank = np.ones(n) * (1 - d) / n

    for page in pages:

        # Get outgoing links
        links = graph[page]

        # Number of outgoing links
        count = len(links)

        # Distribute rank to linked pages
        for link in links:

            index = pages.index(link)

            new_rank[index] += (
                d * rank[pages.index(page)] / count
            )

    # Update rank
    rank = new_rank
```

```python
# 4. DISPLAY FINAL PAGERANK SCORES

print("\nFinal PageRank Scores:")

for i in range(n):
    print(pages[i], ":", round(rank[i], 4))
```

```python
# 5. FIND THE MOST IMPORTANT WEBPAGE

highest_index = np.argmax(rank)

print("\nMost Important Webpage:", pages[highest_index])

print("Highest PageRank Score:", round(rank[highest_index], 4))
```

```python
# 6. DISPLAY RESULT

print("\nInterpretation:")
print("A webpage has a higher PageRank if important")
print("webpages link to it.")
```
