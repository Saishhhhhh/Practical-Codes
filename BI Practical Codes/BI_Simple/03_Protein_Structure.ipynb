```python
import numpy as np

print("Library imported successfully!")
```

```python
# Take protein sequence from user

sequence = input("Enter protein sequence: ")

# Convert to uppercase and remove spaces

sequence = sequence.upper().replace(" ", "")

print("Protein Sequence:", sequence)

print("Protein Length:", len(sequence), "amino acids")
```

```python
def validate_sequence(sequence):

    valid_amino_acids = "ACDEFGHIKLMNPQRSTVWY"

    for amino_acid in sequence:

        if amino_acid not in valid_amino_acids:
            return False

    return len(sequence) > 0


# Validate sequence

if validate_sequence(sequence):

    print("Valid Protein Sequence")

else:

    print("Invalid Protein Sequence")
```

```python
# Define template protein sequence

template_sequence = (
    "MKTIIALSYIFCLVFADYK"
    "DDDDKGQRLIFAGKQLEDGRTLSDY"
    "VAVIDALNDSD"
)

template_name = "Template_Protein_1"

print("Template Name:", template_name)

print("Template Sequence:", template_sequence)

print("Template Length:", len(template_sequence))
```

```python
def sequence_identity(target, template):

    length = min(len(target), len(template))

    matches = 0

    for i in range(length):

        if target[i] == template[i]:

            matches += 1

    identity = (matches / length) * 100

    return identity


# Calculate sequence identity

identity = sequence_identity(
    sequence,
    template_sequence
)

print("Sequence Identity:", round(identity, 2), "%")
```

```python
def simple_alignment(target, template):

    length = min(len(target), len(template))

    target_part = target[:length]

    template_part = template[:length]

    match_line = ""

    for i in range(length):

        if target_part[i] == template_part[i]:

            match_line += "|"

        else:

            match_line += "."

    return target_part, match_line, template_part


# Perform alignment

target_part, match_line, template_part = simple_alignment(
    sequence,
    template_sequence
)


print("Target   :", target_part)

print("           ", match_line)

print("Template :", template_part)
```

```python
def predict_secondary_structure(sequence):

    helix_forming = "ALMEKQRH"

    sheet_forming = "VIFYWT"

    structure = ""

    for amino_acid in sequence:

        if amino_acid in helix_forming:

            structure += "H"

        elif amino_acid in sheet_forming:

            structure += "E"

        else:

            structure += "C"

    return structure


# Predict secondary structure

secondary_structure = predict_secondary_structure(sequence)

print("Protein Sequence:")
print(sequence)

print("\nPredicted Secondary Structure:")
print(secondary_structure)
```

```python
if identity >= 50:

    confidence = "High"

elif identity >= 30:

    confidence = "Moderate"

else:

    confidence = "Low"


print("Sequence Identity:", round(identity, 2), "%")

print("Estimated Modeling Confidence:", confidence)
```

```python
print("POSITION  AMINO ACID  STRUCTURE")

print("-" * 40)

for i, amino_acid in enumerate(sequence):

    structure = secondary_structure[i]

    if structure == "H":

        structure_name = "Alpha Helix"

    elif structure == "E":

        structure_name = "Beta Sheet"

    else:

        structure_name = "Coil"


    print(
        "{:8d}  {:11s}  {}".format(
            i + 1,
            amino_acid,
            structure_name
        )
    )
```

```python
print("=" * 50)

print("PROTEIN STRUCTURE PREDICTION RESULTS")

print("=" * 50)

print("Target Protein Length:", len(sequence))

print("Template:", template_name)

print("Sequence Identity:", round(identity, 2), "%")

print("Modeling Confidence:", confidence)

print("Predicted Secondary Structure:", secondary_structure)

print("\nProtein structure analysis completed.")

print("=" * 50)

# MKTLLVAGAAAGTASAAPVLAADKPEMVRRLAEAGVEVVVEPGDNVVAMRDPQT - Valid
# MKTXYZ123 - Invalid
```
