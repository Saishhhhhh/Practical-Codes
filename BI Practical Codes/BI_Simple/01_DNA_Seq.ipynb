```python
# DNA Sequence Analysis

# DNA consists of four nitrogenous bases:
# A = Adenine
# T = Thymine
# G = Guanine
# C = Cytosine

print("DNA SEQUENCE ANALYSIS")
print("---------------------")
```

```python
# Take DNA sequence from the user

sequence = input("Enter DNA sequence: ")

# Convert sequence to uppercase
sequence = sequence.upper().strip()

print("DNA Sequence:", sequence)
print("Sequence Length:", len(sequence))
```

```python
def validate_dna(sequence):

    valid_bases = "ATGC"

    for base in sequence:
        if base not in valid_bases:
            return False

    return len(sequence) > 0


# Validate the sequence

if validate_dna(sequence):
    print("Valid DNA Sequence")
else:
    print("Invalid DNA Sequence")
```

```python
def calculate_gc_content(sequence):

    g_count = sequence.count("G")
    c_count = sequence.count("C")

    gc_count = g_count + c_count

    gc_content = (gc_count / len(sequence)) * 100

    return gc_content


# Calculate GC content

if validate_dna(sequence):

    gc = calculate_gc_content(sequence)

    print("GC Content:", round(gc, 2), "%")

else:
    print("Cannot calculate GC content.")
```

```python
def find_motif(sequence, motif):

    positions = []

    start = 0

    while True:

        position = sequence.find(motif, start)

        if position == -1:
            break

        # Convert 0-based position to 1-based position
        positions.append(position + 1)

        # Move forward by one position
        start = position + 1

    return positions


# Take motif input

motif = input("Enter motif to search: ").upper().strip()

if validate_dna(motif):

    positions = find_motif(sequence, motif)

    if positions:
        print("Motif found at positions:", positions)
        print("Total occurrences:", len(positions))

    else:
        print("Motif not found.")

else:
    print("Invalid motif.")
```

```python
def find_orfs(sequence):

    start_codon = "ATG"
    stop_codons = ["TAA", "TAG", "TGA"]

    orfs = []

    # Check all three reading frames
    for frame in range(3):

        i = frame

        while i <= len(sequence) - 3:

            codon = sequence[i:i+3]

            # Check for start codon
            if codon == start_codon:

                j = i + 3

                # Search for stop codon
                while j <= len(sequence) - 3:

                    stop_codon = sequence[j:j+3]

                    if stop_codon in stop_codons:

                        orf_sequence = sequence[i:j+3]

                        orfs.append({
                            "start": i + 1,
                            "end": j + 3,
                            "length": len(orf_sequence),
                            "stop": stop_codon,
                            "sequence": orf_sequence
                        })

                        break

                    j += 3

            i += 3

    return orfs


# Identify coding regions

if validate_dna(sequence):

    orfs = find_orfs(sequence)

    if orfs:

        print("Number of ORFs found:", len(orfs))

        for number, orf in enumerate(orfs, start=1):

            print("\nORF", number)
            print("Start Position:", orf["start"])
            print("End Position:", orf["end"])
            print("Stop Codon:", orf["stop"])
            print("Length:", orf["length"], "bp")
            print("ORF Sequence:", orf["sequence"])

    else:
        print("No complete ORF found.")

else:
    print("Invalid DNA sequence.")
```

```python
print("\n" + "=" * 40)
print("DNA SEQUENCE ANALYSIS COMPLETED")
print("=" * 40)

print("Original Sequence:", sequence)
print("Total Length:", len(sequence), "bp")

if validate_dna(sequence):
    print("GC Content:", round(calculate_gc_content(sequence), 2), "%")
    print("DNA Sequence: Valid")
else:
    print("DNA Sequence: Invalid")
```
