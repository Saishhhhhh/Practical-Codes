9. Customize visualizations by adding labels, titles, legends, reference thresholds, and appropriate color schemes.

```python
# Cell 1 - Import Required Libraries
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import matplotlib.dates as mdates

print("Matplotlib and data analysis libraries imported successfully!")
```

```python
# Cell 2 - Tasks 1, 2 & 3: Import, Explore Dataset & Identify Relevant Variables
# Task 1: Import dataset
df = pd.read_csv("City_Air_Quality.csv")
df['Date'] = pd.to_datetime(df['Date'])

# Task 2: Explore dataset structure
print("Dataset Shape (Records, Features):", df.shape)
print("\nFirst 5 rows of City Air Quality dataset:")
display(df.head()) if 'display' in globals() else print(df.head())

print("\nDataset Information:")
df.info()

print("\nSummary Statistics of Pollutants and AQI:")
display(df.describe()) if 'display' in globals() else print(df.describe())

# Task 3: Identify relevant variables for visualization
print("=" * 50)
print("        IDENTIFIED VARIABLES FOR VISUALIZATION")
print("=" * 50)
print("Time Index         : Date (Daily observations)")
print("Primary Metric     : AQI (Air Quality Index)")
print("Pollutant Variables: PM2.5, PM10, CO, NO2")
print("Categorical Variable: AQI_Category (Good, Satisfactory, Moderate, Poor, Very Poor)")
print("=" * 50)
```

```python
# Cell 3 - Tasks 4 & 9: Line Plot of Overall AQI Trend Over Time
# Task 4: Time series line plot of AQI
# Task 9: Customization (labels, title, threshold lines, legend, color schemes)
plt.figure(figsize=(12, 5))
plt.plot(df['Date'], df['AQI'], color='#e74c3c', linewidth=2, marker='o', markersize=3, label='Daily AQI')

# Add reference lines for standard air quality thresholds (CPCB standards)
plt.axhline(50, color='#2ecc71', linestyle='--', alpha=0.7, label='Good (<= 50)')
plt.axhline(100, color='#f1c40f', linestyle='--', alpha=0.7, label='Satisfactory (<= 100)')
plt.axhline(200, color='#e67e22', linestyle='--', alpha=0.7, label='Moderate (<= 200)')
plt.axhline(300, color='#c0392b', linestyle='--', alpha=0.7, label='Poor (<= 300)')

plt.title('Overall Air Quality Index (AQI) Trend Over Time', fontsize=14, fontweight='bold', pad=12)
plt.xlabel('Date', fontsize=11)
plt.ylabel('Air Quality Index (AQI)', fontsize=11)
plt.grid(True, linestyle='--', alpha=0.5)
plt.legend(loc='upper right', framealpha=0.9)
plt.gca().xaxis.set_major_formatter(mdates.DateFormatter('%b %Y'))
plt.tight_layout()
plt.show()
```

```python
# Cell 4 - Tasks 5 & 9: Individual Pollutant Levels on Separate Line Plots
# Task 5: Plot individual pollutants (PM2.5, PM10, CO, NO2) on 2x2 grid subplots
fig, axes = plt.subplots(2, 2, figsize=(14, 8), sharex=True)

# 1. PM2.5
axes[0, 0].plot(df['Date'], df['PM2.5'], color='#8e44ad', linewidth=1.8)
axes[0, 0].set_title('PM2.5 Concentration (µg/m³)', fontsize=12, fontweight='bold')
axes[0, 0].set_ylabel('Concentration (µg/m³)')
axes[0, 0].grid(True, linestyle='--', alpha=0.5)

# 2. PM10
axes[0, 1].plot(df['Date'], df['PM10'], color='#2980b9', linewidth=1.8)
axes[0, 1].set_title('PM10 Concentration (µg/m³)', fontsize=12, fontweight='bold')
axes[0, 1].set_ylabel('Concentration (µg/m³)')
axes[0, 1].grid(True, linestyle='--', alpha=0.5)

# 3. CO
axes[1, 0].plot(df['Date'], df['CO'], color='#d35400', linewidth=1.8)
axes[1, 0].set_title('Carbon Monoxide (CO) Concentration (mg/m³)', fontsize=12, fontweight='bold')
axes[1, 0].set_xlabel('Date')
axes[1, 0].set_ylabel('Concentration (mg/m³)')
axes[1, 0].grid(True, linestyle='--', alpha=0.5)

# 4. NO2
axes[1, 1].plot(df['Date'], df['NO2'], color='#27ae60', linewidth=1.8)
axes[1, 1].set_title('Nitrogen Dioxide (NO2) Concentration (µg/m³)', fontsize=12, fontweight='bold')
axes[1, 1].set_xlabel('Date')
axes[1, 1].set_ylabel('Concentration (µg/m³)')
axes[1, 1].grid(True, linestyle='--', alpha=0.5)

for ax in axes.flat:
    ax.xaxis.set_major_formatter(mdates.DateFormatter('%b %d'))

plt.suptitle('Individual Pollutant Level Trends Over Time', fontsize=15, fontweight='bold', y=0.98)
plt.tight_layout()
plt.show()
```

```python
# Cell 5 - Tasks 6 & 9: Bar Plots and Stacked Bar Plots for Time Periods
# Task 6: Compare AQI across months using Bar Plot & Stacked Bar Plot
df['Month'] = df['Date'].dt.strftime('%B %Y')
monthly_aqi = df.groupby('Month', sort=False)['AQI'].mean().round(1)

plt.figure(figsize=(13, 5))

# Subplot 1: Average Monthly AQI Bar Chart
plt.subplot(1, 2, 1)
bars = plt.bar(monthly_aqi.index, monthly_aqi.values, color=['#3498db', '#e67e22', '#e74c3c', '#9b59b6'], edgecolor='black', width=0.55)
plt.title('Average Monthly AQI Comparison', fontsize=12, fontweight='bold')
plt.xlabel('Month', fontsize=11)
plt.ylabel('Mean AQI Value', fontsize=11)
plt.grid(axis='y', linestyle='--', alpha=0.5)
for bar in bars:
    yval = bar.get_height()
    plt.text(bar.get_x() + bar.get_width()/2.0, yval + 2, f'{yval}', ha='center', va='bottom', fontweight='bold')

# Subplot 2: Stacked Bar Chart of AQI Categories Proportion per Month
plt.subplot(1, 2, 2)
category_pct = pd.crosstab(df['Month'], df['AQI_Category'], normalize='index') * 100
order = [c for c in ['Good', 'Satisfactory', 'Moderate', 'Poor', 'Very Poor'] if c in category_pct.columns]
category_pct = category_pct[order]
cat_colors = {'Good': '#2ecc71', 'Satisfactory': '#f1c40f', 'Moderate': '#e67e22', 'Poor': '#e74c3c', 'Very Poor': '#8e44ad'}

bottom = np.zeros(len(category_pct))
for cat in order:
    plt.bar(category_pct.index, category_pct[cat], bottom=bottom, label=cat, color=cat_colors[cat], edgecolor='black', width=0.55)
    bottom += category_pct[cat].values

plt.title('Monthly Percentage of Days by AQI Category', fontsize=12, fontweight='bold')
plt.xlabel('Month', fontsize=11)
plt.ylabel('Percentage of Days (%)', fontsize=11)
plt.legend(loc='upper right', framealpha=0.9)
plt.grid(axis='y', linestyle='--', alpha=0.5)

plt.tight_layout()
plt.show()
```

```python
# Cell 6 - Tasks 7 & 9: Box Plots for Distribution of AQI by Category
# Task 7: Box plots analyzing AQI distribution across categories
plt.figure(figsize=(9, 5))

category_order = [c for c in ['Good', 'Satisfactory', 'Moderate', 'Poor', 'Very Poor'] if c in df['AQI_Category'].unique()]
box_data = [df[df['AQI_Category'] == cat]['AQI'].values for cat in category_order]

box = plt.boxplot(box_data, label=category_order, patch_artist=True, notch=True,
                  boxprops=dict(facecolor='#aed6f1', color='#1b4f72'),
                  medianprops=dict(color='red', linewidth=2),
                  whiskerprops=dict(color='#1b4f72', linewidth=1.5),
                  capprops=dict(color='#1b4f72', linewidth=1.5),
                  flierprops=dict(marker='o', color='red', alpha=0.6))

plt.title('Distribution of AQI Values by Category (Box Plot)', fontsize=13, fontweight='bold')
plt.xlabel('Air Quality Category', fontsize=11)
plt.ylabel('Air Quality Index (AQI)', fontsize=11)
plt.grid(axis='y', linestyle='--', alpha=0.5)
plt.tight_layout()
plt.show()
```

```python
# Cell 7 - Tasks 8 & 9: Scatter Plot & Bubble Chart (Pollutants vs AQI)
# Task 8: Scatter plot and bubble chart exploring relationships
plt.figure(figsize=(10, 6))

# Bubble Chart:
# X-axis: PM2.5 level
# Y-axis: AQI
# Bubble Size: PM10 level (scaled)
# Bubble Color: CO level (using colormap)
scatter = plt.scatter(
    df['PM2.5'],
    df['AQI'],
    s=df['PM10'] * 0.8,
    c=df['CO'],
    cmap='plasma',
    alpha=0.75,
    edgecolors='black',
    linewidth=0.8
)

cbar = plt.colorbar(scatter)
cbar.set_label('Carbon Monoxide (CO in mg/m³)', fontsize=11)

plt.title('Relationship: PM2.5 vs AQI (Bubble Chart)\n(Size = PM10 Concentration, Color = CO Concentration)', fontsize=13, fontweight='bold')
plt.xlabel('PM2.5 Concentration (µg/m³)', fontsize=11)
plt.ylabel('Air Quality Index (AQI)', fontsize=11)
plt.grid(True, linestyle='--', alpha=0.5)
plt.tight_layout()
plt.show()
```
