# ML 3.2 — K-Nearest Neighbours on Social Network Ads

**Objective:** Implement the K-Nearest Neighbours (KNN) algorithm on the Social Network Ads dataset and calculate the confusion matrix, accuracy, error rate, precision and recall.

**Dataset:** Social Network Ads — Kaggle


```python
# Import required libraries
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns

from sklearn.model_selection import train_test_split
from sklearn.preprocessing import StandardScaler
from sklearn.neighbors import KNeighborsClassifier
from sklearn.metrics import (
    confusion_matrix,
    accuracy_score,
    precision_score,
    recall_score,
    classification_report
)
```

## 1. Load the dataset

```python
# Load dataset
# https://www.kaggle.com/datasets/rakeshrau/social-network-ads
df = pd.read_csv("Social_Network_Ads.csv")

print("First 5 rows:")
display(df.head())

print("\nDataset shape:", df.shape)

print("\nDataset information:")
df.info()
```

## 2. Pre-process the dataset

```python
# Check missing values
print("Missing values:")
print(df.isnull().sum())

# Check duplicate rows
print("\nDuplicate rows:", df.duplicated().sum())

# Remove duplicates if any
df = df.drop_duplicates()

# Display descriptive statistics
print("\nDescriptive statistics:")
display(df.describe())
```

## 3. Select features and target

```python
# Features: Age and EstimatedSalary
# Target: Purchased
X = df[["Age", "EstimatedSalary"]]
y = df["Purchased"]

print("Features:")
display(X.head())

print("\nTarget:")
display(y.head())
```

## 4. Split the dataset into training and testing sets

```python
X_train, X_test, y_train, y_test = train_test_split(
    X,
    y,
    test_size=0.25,
    random_state=42,
    stratify=y
)

print("Training data:", X_train.shape)
print("Testing data:", X_test.shape)
```

## 5. Feature Scaling

```python
# KNN uses distance, so scaling is important
scaler = StandardScaler()

X_train_scaled = scaler.fit_transform(X_train)
X_test_scaled = scaler.transform(X_test)

print("Feature scaling completed.")
```

## 6. Implement K-Nearest Neighbours

```python
# Create KNN classifier
knn = KNeighborsClassifier(n_neighbors=5)

# Train the model
knn.fit(X_train_scaled, y_train)

# Make predictions
y_pred = knn.predict(X_test_scaled)

print("KNN model implemented successfully.")
```

## 7. Confusion Matrix

```python
cm = confusion_matrix(y_test, y_pred)

print("Confusion Matrix:")
print(cm)

plt.figure(figsize=(6, 5))
sns.heatmap(
    cm,
    annot=True,
    fmt="d",
    cmap="Blues",
    xticklabels=["Not Purchased", "Purchased"],
    yticklabels=["Not Purchased", "Purchased"]
)

plt.xlabel("Predicted")
plt.ylabel("Actual")
plt.title("KNN Confusion Matrix")
plt.show()
```

## 8. Calculate Accuracy

```python
accuracy = accuracy_score(y_test, y_pred)

print("Accuracy:", accuracy)
print("Accuracy (%):", accuracy * 100)
```

## 9. Calculate Error Rate

```python
error_rate = 1 - accuracy

print("Error Rate:", error_rate)
print("Error Rate (%):", error_rate * 100)
```

## 10. Calculate Precision

```python
precision = precision_score(y_test, y_pred)

print("Precision:", precision)
```

## 11. Calculate Recall

```python
recall = recall_score(y_test, y_pred)

print("Recall:", recall)
```

## 12. Classification Report

```python
print("Classification Report:")
print(classification_report(
    y_test,
    y_pred,
    target_names=["Not Purchased", "Purchased"]
))
```

## 13. Final Results

```python
results = pd.DataFrame({
    "Metric": [
        "Accuracy",
        "Error Rate",
        "Precision",
        "Recall"
    ],
    "Score": [
        accuracy,
        error_rate,
        precision,
        recall
    ]
})

display(results)
```

## Conclusion

The K-Nearest Neighbours algorithm was successfully implemented on the Social Network Ads dataset. The model was trained using Age and Estimated Salary as input features and Purchased as the target variable. Feature scaling was performed because KNN is a distance-based algorithm. The model was evaluated using the confusion matrix, accuracy, error rate, precision and recall.
